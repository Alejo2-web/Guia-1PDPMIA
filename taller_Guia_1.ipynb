### 1. Identificación de Duplicados y Valores Nulos Básicos
Primero, identificamos si existen filas completamente duplicadas en el conjunto de datos y verificamos la presencia de nulos tradicionales (`NaN` o `None`).

```python

cant_duplicados = df.duplicated().sum()
print(f"Cantidad de filas duplicadas identificadas: {cant_duplicados}")
nulos_por_columna = df.isnull().sum()
print("\nValores nulos tradicionales por columna:")
print(nulos_por_columna[nulos_por_columna > 0] if nulos_por_columna.sum() > 0 else "No se encontraron nulos tradicionales (NaN/None).")
```

### 2. Eliminación de Registros Duplicados
Si existen registros duplicados, procedemos a eliminarlos para no sesgar nuestros análisis y modelos.

```python
print(f"Forma original del dataset: {df.shape}")
df = df.drop_duplicates().reset_index(drop=True)
print(f"Forma del dataset después de eliminar duplicados: {df.shape}")
```

### 3. Identificación y Tratamiento de Valores Implícitos ('unknown')
En este dataset de telemarketing bancario, los valores nulos o faltantes no vienen representados como `NaN`, sino con la etiqueta categórica `'unknown'`. Vamos a cuantificarlos y tratarlos.

```python
import numpy as np
for col in df.select_dtypes(include=['object']).columns:
    cant_unknown = (df[col] == 'unknown').sum()
    if cant_unknown > 0:
        pct = (cant_unknown / len(df)) * 100
        print(f"Columna '{col}': {cant_unknown} valores 'unknown' ({pct:.2f}%)")
```

### 4. Ajuste y Optimización de Tipos de Datos
Convertir las variables de tipo texto (`object`) a categorías (`category`) optimiza el rendimiento y prepara los datos para los algoritmos.

```python
columnas_objeto = df.select_dtypes(include=['object']).columns
for col in columnas_objeto:
    df[col] = df[col].astype('category')


print(df.dtypes)
```

### 1. Distribución de la Variable Objetivo (`y`)
Este gráfico permite analizar qué tan balanceadas o desbalanceadas están las clases en el conjunto de datos (suscripción exitosa vs. no exitosa).

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='y', data=df, palette='Set2', hue='y', legend=False)
plt.title('Distribución de Suscripción al Depósito (y)')
plt.xlabel('¿Suscrito?')
plt.ylabel('Cantidad de Clientes')
plt.show()
```

El Bloque Económico (Relación Muy Fuerte)
euribor3m (tasas de interés) - nr.employed (empleados) - emp.var.rate (tasa de empleo): Tienen una correlación casi perfecta (entre 0.91 y 0.97). Esto significa que cuando la economía está fuerte (sube el empleo), las tasas de interés también suben de inmediato. Aportan casi la misma información.

### 2. Distribución de la Edad según la Suscripción
Este análisis ayuda a comprender a qué edades es más probable que los clientes se suscriban al depósito.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', hue='y', multiple='stack', palette='Set2', kde=True)
plt.title('Distribución de Edad según Suscripción')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.show()
```

Campañas de Contacto (Relación Moderada)
pdays (días desde el último contacto) y previous (contactos anteriores): Tienen una relación inversa (-0.58). Cuanto más contacto se ha tenido con un cliente en el pasado, menos tiempo ha transcurrido desde la última vez que se le llamó.
euribor3m (tasas de interés) y previous (contactos anteriores): Tienen una relación negativa (-0.58). En épocas con tasas de interés más bajas, se realizaron más campañas y contactos previos.

### 3. Suscripción por Tipo de Trabajo (`job`)
Este gráfico permite identificar cuáles sectores u ocupaciones tienen mayor volumen de contactos y una mejor tasa de conversión.

```python
plt.figure(figsize=(10, 6))
sns.countplot(y='job', hue='y', data=df, palette='Set2', order=df['job'].value_counts().index)
plt.title('Suscripción por Tipo de Trabajo')
plt.xlabel('Cantidad de Clientes')
plt.ylabel('Trabajo')
plt.show()
```

Edad y Comportamiento (Sin Relación Directa)
La edad (age) no tiene apenas relación matemática directa con el resto de variables numéricas. Su comportamiento (como la preferencia de jóvenes y jubilados) se ve de forma visual en los gráficos de distribución, no de forma lineal.

### 4. Matriz de Correlación de Variables Numéricas
Este mapa de calor muestra el grado de relación lineal entre las variables numéricas y los indicadores macroeconómicos.

```python
plt.figure(figsize=(10, 8))
numeric_cols = ['age', 'duration', 'campaign', 'pdays', 'previous', 'emp.var.rate', 'cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']
corr_matrix = df[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', cbar=True, annot_kws={'size': 9})
plt.title('Matriz de Correlación de Variables Numéricas')
plt.tight_layout()
plt.show()
```
